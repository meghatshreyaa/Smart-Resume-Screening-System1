#Import Libraries

```python
import re
import PyPDF2
import pandas as pd

from google.colab import files
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

print("Libraries imported successfully!")
```

#Upload Resume

```python
uploaded = files.upload()

file_name = list(uploaded.keys())[0]

print("Uploaded file:", file_name)
```

#Extract Text from Resume

```python
def extract_text_from_pdf(file_name):
    text = ""

    with open(file_name, "rb") as file:
        pdf_reader = PyPDF2.PdfReader(file)

        for page in pdf_reader.pages:
            page_text = page.extract_text()

            if page_text:
                text += page_text + "\n"

    return text


resume_text = extract_text_from_pdf(file_name)

print("Resume text extracted successfully!")
print("\n--- Resume Preview ---\n")
print(resume_text[:3000])
```

#Enter Job Description

```python
job_description = """
We are looking for a Python Developer with experience in Python,
SQL, Machine Learning, Pandas, NumPy, Flask and Data Analysis.

The candidate should have good problem-solving and communication skills.
Experience with Scikit-learn, Git and GitHub is preferred.
"""

print(job_description)
```

#Clean and Preprocess Text

```python
def clean_text(text):

    text = text.lower()

    # Remove email addresses
    text = re.sub(r'\S+@\S+', ' ', text)

    # Remove URLs
    text = re.sub(r'http\S+|www\S+', ' ', text)

    # Remove special characters
    text = re.sub(r'[^a-zA-Z0-9\s]', ' ', text)

    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text)

    return text.strip()


clean_resume = clean_text(resume_text)
clean_job = clean_text(job_description)

print("Text preprocessing completed!")
```

#Calculate Resume Match Score

```python
documents = [
    clean_resume,
    clean_job
]

vectorizer = TfidfVectorizer(
    stop_words="english",
    ngram_range=(1, 2)
)

tfidf_matrix = vectorizer.fit_transform(documents)

similarity = cosine_similarity(
    tfidf_matrix[0:1],
    tfidf_matrix[1:2]
)

match_score = round(similarity[0][0] * 100, 2)

print("Resume Match Score:", match_score, "%")
```

#Extract Skills

```python
skills_list = [
    "python",
    "java",
    "c",
    "c++",
    "sql",
    "mysql",
    "mongodb",
    "html",
    "css",
    "javascript",
    "react",
    "node.js",
    "flask",
    "django",
    "machine learning",
    "deep learning",
    "data science",
    "artificial intelligence",
    "power bi",
    "excel",
    "tableau",
    "pandas",
    "numpy",
    "scikit-learn",
    "tensorflow",
    "keras",
    "nlp",
    "git",
    "github",
    "aws",
    "azure",
    "cloud computing",
    "communication",
    "leadership",
    "problem solving",
    "data analysis"
]


def extract_skills(text):

    text = text.lower()

    found_skills = []

    for skill in skills_list:

        if skill.lower() in text:
            found_skills.append(skill)

    return sorted(set(found_skills))


resume_skills = extract_skills(resume_text)

job_skills = extract_skills(job_description)

matching_skills = sorted(
    set(resume_skills) & set(job_skills)
)

missing_skills = sorted(
    set(job_skills) - set(resume_skills)
)


print("Resume Skills:")
print(resume_skills)

print("\nJob Required Skills:")
print(job_skills)

print("\nMatching Skills:")
print(matching_skills)

print("\nMissing Skills:")
print(missing_skills)
```

#Candidate Recommendation

```python
def get_recommendation(score):

    if score >= 75:
        return "Strong Match - Shortlist"

    elif score >= 50:
        return "Moderate Match - Review"

    else:
        return "Low Match - Reject"


recommendation = get_recommendation(match_score)

print("=" * 50)
print("SMART RESUME SCREENING RESULT")
print("=" * 50)

print("Resume:", file_name)

print("Match Score:", match_score, "%")

print("Recommendation:", recommendation)

print("Matching Skills:", matching_skills)

print("Missing Skills:", missing_skills)

print("=" * 50)
```

#Display Final Result

```python
result = pd.DataFrame({
    "Resume": [file_name],
    "Match Score (%)": [match_score],
    "Recommendation": [recommendation],
    "Matching Skills": [", ".join(matching_skills)],
    "Missing Skills": [", ".join(missing_skills)]
})

display(result)
```

#Download Result

```python
output_file = "resume_screening_result.csv"

result.to_csv(
    output_file,
    index=False
)

files.download(output_file)

print("Result downloaded successfully!")
```
